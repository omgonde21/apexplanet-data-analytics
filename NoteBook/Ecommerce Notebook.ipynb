```python
import os

os.makedirs('../data/raw', exist_ok=True)
os.makedirs('../data/processed', exist_ok=True)
os.makedirs('../reports', exist_ok=True)

print("✅ Folder structure ready ")
```

```python
import pandas as pd

df = pd.read_csv('data.csv', encoding='ISO-8859-1')

```

```python

df = df.dropna(subset=['CustomerID'])
df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)]

```

```python

df['TotalAmount'] = df['Quantity'] * df['UnitPrice']

```

```python

df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

print("✅ Data cleaned")
print("Final rows  count:", len(df))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns


top_5 = df.groupby('Description')['Quantity'].sum().sort_values(ascending=False).head(5)


plt.figure(figsize=(10, 4))
sns.barplot(x=top_5.values, y=top_5.index, palette="Blues_r")
plt.title("Top 5 Best Selling Products")
plt.xlabel("Total Quantity Sold")
plt.show()

```

```python
import os

# 1. Folder create karo (agar pehle se nahi bana hai)
os.makedirs('cleaned_data', exist_ok=True)

# 2. Data ko us folder ke andar save karo
df.to_csv('cleaned_data/cleaned_ecommerce.csv', index=False)

print("✅ File 'cleaned_data/cleaned_ecommerce.csv' File Saved")
```
